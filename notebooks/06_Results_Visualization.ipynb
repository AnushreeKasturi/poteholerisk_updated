```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Risk_Level",
    data=accident_df,
    order=["Low","Moderate","High","Critical"]
)

plt.title("Distribution of Bayesian Risk Levels")

plt.xlabel("Risk Level")

plt.ylabel("Number of Accidents")

plt.show()
```

```python
weather_risk = (

    accident_df
    .groupby("Weather")["Final_Risk"]
    .mean()
    .sort_values(ascending=False)

)

plt.figure(figsize=(10,6))

weather_risk.plot(kind="bar")

plt.title("Average Final Risk by Weather")

plt.ylabel("Average Risk")

plt.xticks(rotation=45)

plt.show()
```

```python
road_risk = (

    accident_df
    .groupby("Road_Condition")["Final_Risk"]
    .mean()
    .sort_values(ascending=False)

)

plt.figure(figsize=(10,6))

road_risk.plot(kind="bar")

plt.title("Average Final Risk by Road Condition")

plt.ylabel("Average Risk")

plt.xticks(rotation=60, ha="right")
plt.tight_layout()

plt.show()
```

```python
cause_risk = (

    accident_df
    .groupby("Main_Cause")["Final_Risk"]
    .mean()
    .sort_values(ascending=False)

)

plt.figure(figsize=(8,5))

cause_risk.plot(kind="bar")

plt.title("Average Final Risk by Main Cause")

plt.ylabel("Average Risk")

plt.xticks(rotation=20)

plt.show()
```

```python
plt.figure(figsize=(8,6))

corr = accident_df[

    [

        "Prior_Probability",

        "Posterior_Probability",

        "Traffic_Exposure",

        "Final_Risk"

    ]

].corr()

sns.heatmap(

    corr,

    annot=True,

    cmap="coolwarm",

    square=True

)

plt.title("Correlation Between Bayesian Variables")

plt.show()
```
