```python
bengaluru_accidents = accident_df[
    accident_df["DISTRICTNAME"].isin(
        ["Bengaluru City", "Bengaluru Dist"]
    )
]

print(bengaluru_accidents.shape)
```

```python

print("Total Columns:", len(bengaluru_accidents.columns))
print()

for i, col in enumerate(bengaluru_accidents.columns, start=1):
    print(f"{i}. {col}")
```

```python
# Explore categorical columns

categorical_columns = [
    "Weather",
    "Road_Condition",
    "Surface_Condition",
    "Road_Type",
    "Road_Character",
    "Collision_Type",
    "Main_Cause",
    "Severity"
]

for col in categorical_columns:
    print("\n" + "="*60)
    print(col)
    print("="*60)
    print(bengaluru_accidents[col].value_counts())
```
